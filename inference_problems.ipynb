1. A random sample has
   $$
   \bar{X}=52,\qquad s=10,\qquad n=100.
   $$
   Compute the estimated standard error of $\bar{X}$. Test
   $$
   H_0:\mu=50
   \qquad \text{vs.} \qquad
   H_1:\mu\ne 50
   $$
   at significance levels $\alpha=.10,.05,.01$ using the large-sample $z$-statistic. These correspond to 90%, 95%, and 99% confidence levels. Construct a 95% confidence interval for $\mu$.

Q1

The estimated standard error is
   $$
   \widehat{SE}(\bar{X})=\frac{s}{\sqrt{n}}
   =\frac{10}{\sqrt{100}}=1.
   $$

   The test statistic is
   $$
   z=\frac{\bar{X}-\mu_0}{\widehat{SE}(\bar{X})}
   =\frac{52-50}{1}=2.
   $$

   - At $\alpha=.10$, I reject $H_0$ because 2 > 1.645.
   - At $\alpha=.05$, I reject $H_0$ because 2 > 1.96.
   - At $\alpha=.01$, I fail to reject $H_0$ because 2 < 2.576.

   The 95% confidence interval for $\mu$ is
   $$
   52\pm1.96(1)=(50.04,\ 53.96).
   $$

   The interval excludes 50, agreeing with rejecting $H_0$ at $\alpha=.05$

For the following data and scenarios (7-11):

   - a. Generate at least $R=2{,}000$ nonparametric bootstrap estimates with a reproducible RNG. Report the seed and $R$. For two groups, resample independently within groups; for correlation, resample paired rows.
   - b. Estimate the standard error $\widehat{SE}(\hat\theta)$ with the sample standard deviation of the bootstrap estimates (using denominator $R-1$, or `ddof=1`).
   - c. Construct centered-Normal 90%, 95%, and 99% CIs: $\hat\theta \pm z^*\widehat{SE}(\hat\theta)$.
   - d. Test the stated null using $z=(\hat\theta-\theta_0)/\widehat{SE}(\hat\theta)$ and $p=2\Phi(-|z|)$. Decide at $\alpha=.10,.05,.01$.
   - e. Confirm that every test decision agrees with whether the matching CI contains $\theta_0$.
   - f. Interpret the estimate, interval, and test in context. These are observational data: do not turn association into causation.

Use unrounded estimates and standard errors when computing intervals and tests. If a p-value is close to a decision threshold, increase $R$ (for example, to $100{,}000$, using batches if needed) to reduce Monte Carlo error. Small seed-dependent differences in results or borderline test decisions are acceptable when the procedure is correct and reproducible and you discuss the remaining simulation uncertainty. In particular, Problem 10 may be close to the 5% threshold.

```python
# My Setup Code

import numpy as np
import pandas as pd
from scipy.stats import norm

ames = pd.read_csv('./data/ames_prices.csv')
metabric = pd.read_csv('./data/metabric.csv')
nhanes = pd.read_csv('./data/nhanes_data_17_18.csv', low_memory=False)

def report_inference(estimate, results, null_value):
    standard_error = np.std(results, ddof=1)
    z = (estimate - null_value) / standard_error
    p_value = 2 * norm.cdf(-abs(z))

    print(f'Estimate: {estimate:.6f}')
    print(f'Bootstrap standard error: {standard_error:.6f}')
    print(f'z: {z:.6f}')
    print(f'Two-sided p-value: {p_value:.6g}')

    for confidence, alpha, critical in [
        (0.90, 0.10, 1.645),
        (0.95, 0.05, 1.96),
        (0.99, 0.01, 2.576)
    ]:
        lower = estimate - critical * standard_error
        upper = estimate + critical * standard_error

        reject = p_value < alpha
        contains_null = lower <= null_value <= upper
        decision = 'Reject H0' if reject else 'Fail to reject H0'

        print(f'\n{confidence:.0%} CI: ({lower:.6f}, {upper:.6f})')
        print(f'At alpha = {alpha:.2f}: {decision}')
        print(f'CI contains null value: {contains_null}')
        print(f'Test and CI agree: {reject == (not contains_null)}')
```

7. For `./data/ames_prices.csv`, define an indicator equal to 1 when `Central.Air == 'Y'`. Estimate the proportion of homes with central air and test $H_0:p=.90$. Treat a proportion as the mean of a binary variable and complete the full inference routine.

```python
# My Q7 Code

seed = 100
rng = np.random.default_rng(seed)
R = 10_000

central_air = (ames['Central.Air'].dropna() == 'Y').astype(int).to_numpy()
estimate = central_air.mean()

results = []
for t in range(R):
    idx = rng.integers(0, len(central_air), len(central_air))
    results.append(central_air[idx].mean())

results = np.array(results)

print(f'Seed: {seed}, R: {R}')
print(f'Number of homes: {len(central_air)}')
report_inference(estimate, results, 0.90)
```

8. For `./data/ames_prices.csv`, estimate the Pearson correlation between `arcsinh(area)` and `arcsinh(price)`. Bootstrap paired rows, construct all three CIs, and test $H_0:\rho=0$. Explain what would go wrong if area and price were resampled separately.

```python
# My Q8 Code

seed = 100
rng = np.random.default_rng(seed)
R = 10_000

paired_data = ames[['area', 'price']].dropna()
transformed = np.arcsinh(paired_data.to_numpy())
estimate = np.corrcoef(transformed[:, 0], transformed[:, 1])[0, 1]

results = []
for t in range(R):
    idx = rng.integers(0, len(transformed), len(transformed))
    sample = transformed[idx]

    correlation = np.corrcoef(sample[:, 0], sample[:, 1])[0, 1]
    results.append(correlation)

results = np.array(results)

print(f'Seed: {seed}, R: {R}')
print(f'Number of paired observations: {len(transformed)}')
report_inference(estimate, results, 0)
```

10. For `./data/metabric.csv`, create `deceased = 1` when `Overall Survival Status == '1:DECEASED'`. Estimate the difference in deceased proportions between `Chemotherapy == 'YES'` and `'NO'`: YES minus NO. Test $H_0:\Delta=0$. State at least two plausible confounders visible in the dataset.

```python
# My Q10 Code

seed = 100
rng = np.random.default_rng(seed)
R = 100_000

df = metabric[['Chemotherapy', 'Overall Survival Status']].dropna().copy()
df['deceased'] = (df['Overall Survival Status'] == '1:DECEASED').astype(int)

yes = df.loc[df['Chemotherapy'] == 'YES', 'deceased'].to_numpy()
no = df.loc[df['Chemotherapy'] == 'NO', 'deceased'].to_numpy()
estimate = yes.mean() - no.mean()

results = []
for t in range(R):
    idx_yes = rng.integers(0, len(yes), len(yes))
    idx_no = rng.integers(0, len(no), len(no))

    difference = yes[idx_yes].mean() - no[idx_no].mean()
    results.append(difference)

results = np.array(results)

print(f'Seed: {seed}, R: {R}')
print(f'YES sample size: {len(yes)}')
print(f'NO sample size: {len(no)}')
print(f'YES deceased proportion: {yes.mean():.6f}')
print(f'NO deceased proportion: {no.mean():.6f}')
report_inference(estimate, results, 0)
```

For $H_0:\Delta=0$, z = -1.945873 and the p-value is 0.05167. I reject $H_0$ at $\alpha=.10$, but fail to reject it at $\alpha=.05$ and $\alpha=.01$.
